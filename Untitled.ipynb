```python
from xml.etree.ElementTree import parse, dump

def getAccessionGOTerms(xmlFile, accession):
    tree = parse(xmlFile)
    root = tree.getroot()
    ns = '{http://uniprot.org/uniprot}'
    
    match_go_terms = []
    proteins = root.findall(ns + 'entry')
    for entry in proteins:
        accessions = entry.findall(ns+"accession")
        current_accessions = [ acc.text for acc in accessions ]
        if not accession in current_accessions:
            continue
        goTerms = entry.findall('{http://uniprot.org/uniprot}dbReference[@type="GO"]')
        #goTerms = xmlEntry.findall(ns +'dbReference[@type="GO"]')
        for goT in goTerms:
            gID   = goT.attrib['id']
            gName = goT.find(ns +'property[@type="term"]').attrib['value']
            match_go_terms.append((gID, gName))
        break
    return match_go_terms
print(getAccessionGOTerms("./data/uniprot-proteome_UP000000625.xml", "P0A8V6"))

a = dict(getAccessionGOTerms("./data/uniprot-proteome_UP000000625.xml", "P0A8V6"))
print(a)
```

```python
def dictionnaire_termes_go(chemin_fichier_xml,prot_surrabondantes):
    dictionnaire_termes_GO = {}
    for i in prot_surrabondantes :
        go_list = getAccessionGOTerms(chemin_fichier_xml, i)
    
        for go_id, go_name in go_list: 
            if go_id not in dictionnaire_termes_GO:
                dictionnaire_termes_GO[go_id] = {
                    'ID': go_id,
                    'name': go_name,
                    'carried_by':[]
                }
                
            if i not in dictionnaire_termes_GO[go_id]['carried_by']:
                dictionnaire_termes_GO[go_id]['carried_by'].append(i)
    return dictionnaire_termes_GO
```

```python
import json

with open("/Users/maelebedel/Desktop/INSA/Cours/5A/Proteomique/tp-proteomics/data/EColiK12_GOcounts.json", "r", encoding="utf-8") as f:
    data = json.load(f)   # data = dict ou list selon le contenu

print(type(data))
print(data)
```

```python
from scipy.stats import hypergeom
# hypergeom(M, n, N)
# M=N (taille de la population)
# n=K (nb de succès dans la population)
# N=n (nb de tirages) 

alpha = 0.05

N = 1746
n = 23
go_terms = data["go_terms"]

results = []
for go_id, info in dictionnaire_termes_GO.items():
    k = len(set(info.get("carried_by", []))) # nombre de surabondantes portant ce GO
    K = int(go_terms.get(go_id, {}).get("count",0)) # nombre dans le fond portant ce GO
    
    pval = hypergeom.sf(k-1, N, K, n) # P(X>= k), Hypergeom(N, K, n)

    results.append({
        "go_id": go_id,
        "name": info.get("name"),
        "k": k, "K": K,
        "pval": pval
    })

    results.sort(key=lambda r: r["pval"])

for r in results:
    print(f'{r["go_id"]}\t{r["name"]}\tk={r["k"]}\tK={r["K"]}\tp={r["pval"]}')
```

```python
significant = []
for r in results : 
    if r["pval"] < alpha:
        significant.append(r)

significant.sort(key=lambda r: r["pval"])

for r in significant:
    print(f'{r["go_id"]}\t{r["name"]}\tk={r["k"]}\tK={r["K"]}\tp={r["pval"]}')
```

```python
import re
def mitab_parser(mitab:str, uniprot_ac:list[str])->list[str]:
    """ Extract from provided file (path), lines featurin at least one interactor present in uniprot_ac provided list
        Returns: a list of mitab records split over tabulation.
    """
    hits = []
    with open(mitab, 'r') as fp:
        for line in fp:
            line = line.split()
            for proteinfield in line[0:2]: # Test column one and two
                _ = re.match("uniprotkb:(\\S+)", proteinfield) # Extract uniprot identifier
                x= _[1] if _ else None
                                
                if x in uniprot_ac: # Check uniprot identifier is in provided ac list
                    hits.append(line)
                    break

    print(f"Filtered {len(hits)} interactions")
    return hits

res = mitab_parser('/Users/maelebedel/Desktop/INSA/Cours/5A/Proteomique/tp-proteomics/data/proteins.mitab',prot_surrabondantes)
```
